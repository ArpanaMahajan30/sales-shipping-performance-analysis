📊 Sales & Shipping Performance Analysis

Objective:
Analyze sales performance across product categories and evaluate shipping efficiency using order and ship dates.

🔹 Category-wise Sales Analysis

Grouped data by product category and calculated total sales.

Technology category generated the highest revenue, followed by Furniture and Office Supplies.

Sales distribution indicates that Technology products are the primary revenue drivers.

Insight:
Focusing marketing and inventory strategies on the Technology category can maximize revenue.

🔹 Shipping Duration Analysis

Calculated shipping days as the difference between ship date and order date.

Average shipping time is approximately 4 days.

75% of orders are delivered within 5 days, indicating efficient logistics operations.

Maximum shipping duration observed was 7 days, with no extreme delays.

Insight:
Shipping performance is consistent and reliable, contributing to better customer satisfaction.

✅ Conclusion

Technology category contributes the highest sales.

Shipping operations are stable with predictable delivery timelines.

The analysis provides actionable insights for improving sales strategy and operational efficiency.
